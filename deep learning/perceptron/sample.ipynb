```python
from sklearn.datasets import make_classification
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x,y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,n_classes=2,n_clusters_per_class=1,random_state=50,hypercube=False,class_sep=10)
```

```python
plt.scatter(x[:,0],x[:,1],c=y,s=100)
```

```python

def perceptron(x,y):
    x=np.insert(x,0,1,axis=1)
    weights=np.ones(x.shape[1])
    lr=0.1
    for i in range(1000):
        j=np.random.randint(0,100)
        y_hat=step(np.dot(x[j],weights))
        weights=weights+lr*(y[j]-y_hat)*x[j]
    return weights[0],weights[1:]
def step(z):
    return 1 if z>0 else 0        
```

```python
np.dot(np.insert(x[41],0,1),[1,1,1])
```

```python
step(np.dot(np.insert(x[41],0,1),[1,1,1]))
```

```python
intercept_,coef_=perceptron(x,y)
```

```python
intercept_,coef_
```

```python
m=coef_[0]/coef_[1]
b=intercept_/coef_[1]
```

```python
b
```

```python
x_input= np.linspace(-3,3,100)
y_input = m*x_input+b
```

```python
plt.plot(x_input,y_input,color="red",linewidth=3)
plt.scatter(x[:,0],x[:,1],c=y)
```

```python
def peceptron(x,y):
    w1=w2=b=1
    lr=0.1
    for j in range(1000):
        for i in range(x.shape[0]):
            z=w1*x[i][0]+w2*x[i][1]+b
            if z*y[i]<0:
                w1=w1+lr*y[i]*x[i][0]
                w2=w2+lr*y[i]*x[i][1]
                b=b+lr*y[i]
    return w1,w2,b
```

```python
w1,w2,b= peceptron(x,y)
```

```python
b
```

```python
m=-(w1/w2)
c=-(b/w2)
```

```python
x_input= np.linspace(-3,3,100)
y_input = m*x_input+c
```

```python
plt.plot(x_input,y_input,color="red",linewidth=3)
plt.scatter(x[:,0],x[:,1],c=y)
```
