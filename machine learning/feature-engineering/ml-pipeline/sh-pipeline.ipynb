# titanic without using pipeline

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import accuracy_score
```

```python
df= pd.read_csv("../../understanding-data/train.csv")
df.sample(5)
```

```python
df.drop(columns=["PassengerId", "Name", "Ticket", "Cabin"],inplace=True)
df.sample(5)
```

```python
x_train, x_test, y_train, y_test = train_test_split(df.drop(columns=["Survived"]),df["Survived"],test_size=0.2,random_state=42)
```

```python
#applying imputation
si_age=SimpleImputer()
si_embarked =SimpleImputer(strategy="most_frequent")

x_train_age=si_age.fit_transform(x_train[["Age"]])
x_train_embarked=si_embarked.fit_transform(x_train[["Embarked"]])

x_test_age=si_age.fit_transform(x_test[["Age"]])
x_test_embarked=si_embarked.fit_transform(x_test[["Embarked"]])
```

```python
x_train_age
```

```python
# applying one hot on sex and embarked
ohe_sex=OneHotEncoder(sparse_output=False,handle_unknown="ignore")
ohe_embarked=OneHotEncoder(sparse_output=False,handle_unknown="ignore")

x_train_sex=ohe_sex.fit_transform(x_train[["Sex"]])
x_train_embarked=ohe_embarked.fit_transform(x_train_embarked)

x_test_sex=ohe_sex.fit_transform(x_test[["Sex"]])
x_test_embarked=ohe_embarked.fit_transform(x_test_embarked)
```

```python
x_train_rem =x_train.drop(columns=["Embarked", "Sex", "Age"])
x_test_rem =x_test.drop(columns=["Embarked", "Sex", "Age"])
```

```python
x_train_rem
```

```python
x_train_transform = np.concatenate((x_train_age,x_train_embarked,x_train_sex,x_train_rem),axis=1)
x_test_transform = np.concatenate((x_test_age,x_test_embarked,x_test_sex,x_test_rem),axis=1)
```

```python
x_train_transform.shape
```

```python
clf = DecisionTreeClassifier()
clf.fit(x_train_transform,y_train)
```

```python
y_pred = clf.predict(x_test_transform)
y_pred
```

```python
print(accuracy_score(y_pred,y_test))
```

# now we will do the same using piplelines

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.feature_selection import SelectKBest,chi2
from sklearn.preprocessing import MinMaxScaler
```

```python
df
```

```python
x_train, x_test, y_train, y_test = train_test_split(df.drop(columns=["Survived"]),df["Survived"],test_size=0.2,random_state=42)
```

```python
trf1=ColumnTransformer([
("impute_age",SimpleImputer(),[2]),
("impute_embaked",SimpleImputer(strategy="most_frequent"),[6])
],remainder="passthrough")
```

```python
# one hot encoding transformer

trf2=ColumnTransformer([
    ("ohe_sex_embarked",OneHotEncoder(sparse_output=False,handle_unknown="ignore"),[1,6])
],remainder="passthrough")
```

```python
# scaling 
trf3 = ColumnTransformer([
    ("scaling",MinMaxScaler(),slice(0,10))
])
```

```python
# feature sleaction
trf4 = SelectKBest(score_func=chi2,k=8)
```

```python
# model training 
trf5 = DecisionTreeClassifier()
```

```python
pipe=Pipeline([
    ("trf1",trf1),
    ("trf2",trf2),
    ("trf3",trf3),
    # ("trf4",trf4),
    ("trf5",trf5)
])
```

```python
pipe=make_pipeline(trf1,trf2,trf3,trf4,trf5)
```

```python
pipe.fit(x_train,y_train)
```

```python
pipe.named_steps
```

```python
pipe.named_steps["trf1"].transformers_[0][1].statistics_
```

```python
y_pred =pipe.predict(x_test)
y_pred
```

```python
accuracy_score(y_test,y_pred)
```

# cross_val_score

```python
from sklearn.model_selection import cross_val_score
```

```python
cross_val_score(pipe,x_train,y_train,cv=5,scoring="accuracy").mean()
```

# grid search

```python
params={
    'trf5__max_depth':[1,2,3,4,5,None]
}
```

```python
from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(pipe, params, cv=5, scoring='accuracy')
grid.fit(x_train, y_train)
```
