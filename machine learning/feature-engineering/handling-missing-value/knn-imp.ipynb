```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import KNNImputer, SimpleImputer

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score
```

# KNN Imputer

```python
df = pd.read_csv( "../../understanding-data/train.csv", usecols=["Age","Fare", "Survived","Pclass"] )
```

```python
df.isnull().mean()
```

```python
x = df.drop(columns=["Survived"])
y=df["Survived"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
knn=KNNImputer(add_indicator=True,n_neighbors=3)
x_train_trf=knn.fit_transform(x_train)
x_test_trf = knn.fit_transform(x_test)
x_train.sample()
```

```python
pd.DataFrame(x_train_trf, columns=["Pclass","Age","Fare","age_na"]).sample(10)
```

```python
lr=LogisticRegression()
lr.fit(x_train_trf,y_train)
y_pred = lr.predict(x_test_trf)

accuracy_score(y_pred,y_test)
```
