```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv("student_clustering.csv")
```

```python
df
```

```python
sns.scatterplot(data=df, x="cgpa", y="iq")
```

```python
from sklearn.cluster import KMeans
```

```python
wcss=[]
for i in range(1,11):
    km=KMeans(n_clusters=i)
    km.fit(df)
    wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,11),wcss)
plt.show()
```

```python
X=df.iloc[:,:].values
```

```python
km=KMeans(n_clusters=4)
y_mean=km.fit_predict(X)
```

```python
y_mean
```

```python
X[y_mean==1]
```

```python
plt.scatter(X[y_mean==0,0],X[y_mean==0,1],c="blue")
plt.scatter(X[y_mean==1,0],X[y_mean==1,1],c="red")
plt.scatter(X[y_mean==2,0],X[y_mean==2,1],c="green")
plt.scatter(X[y_mean==3,0],X[y_mean==3,1],c="yellow")
```
